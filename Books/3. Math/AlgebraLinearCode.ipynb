### 1.1 Weights

In a neural network, weights do not exist in isolation. For each layer, all the weights are arranged in a single matrix, conventionally called W. What makes this elegant is the geometric interpretation: each column of this matrix carries the weights of a single neuron in that layer. That is, if the layer has 4 neurons, the matrix has 4 columns, and each column concentrates everything that neuron "knows how to look at" in the input.

The lines, in turn, correspond to the input features. If the input vector has 3 dimensions, each column of W will have 3 elements, one for each feature. When you multiply W by the input vector x, what happens is that each neuron calculates its own inner product with x simultaneously, without any explicit loop. The entire layer processes the batch at once with a single linear algebra operation.

Now when there is a second layer, it does not share the same matrix. It requires a new W2 matrix, with completely different dimensions. The number of W2 lines needs to be equal to the number of neurons in the previous layer, because the output of layer 1 becomes the input of layer 2. The number of columns of W2 defines how many neurons layer 2 has. Each new layer is literally a new linear transformation, represented by a new independent matrix.

This is why stacking layers in a neural network is equivalent to composing linear transformations separated by nonlinearities. Each matrix W learns to project the data into a new space, and the depth of the network is the number of such chained projections.
