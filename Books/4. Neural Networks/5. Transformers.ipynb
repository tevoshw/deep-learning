# QKV (Pesos)


As matrizes QKV representam

Q = Perguntas
K = Respostas
V = Valores

Aonde perguntas == respostas  == valores (qkv) e o shape delas são (embedding(d_model), qkv)
Aonde multiplicaremos pelo x (tokens, d_model) pelo (d_model, qkv) e assim ficando (tokens, qkv), se fosse ao contrario daria 


## QKV

- Cada linha de Q (tokens, qkv) representa um token e cada coluna uma pergunta
- Cada linha de K (tokens, qkv) representa um token e cada coluna uma resposta para a pergunta
- Cada linha de V (tokens, qkv) representa um token e cada coluna um valor
