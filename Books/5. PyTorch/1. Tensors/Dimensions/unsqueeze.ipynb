# Unsqueeze add a new dimension 

```python
import torch
x = torch.rand(5, 10) 
print(f'Shape of X: {x.shape}')
```

#### 1. Add in the front

```python
new_dim_front = x.unsqueeze(0)
print(f'Shape of X: {new_dim_front.shape}')
```

#### 1. Add in the mid

```python
new_dim_mid = x.unsqueeze(1)
print(f'Shape of X: {new_dim_mid.shape}')
```

#### 1. Add in the back

```python
new_dim_back = x.unsqueeze(-1)
print(f'Shape of X: {new_dim_back.shape}')
```
