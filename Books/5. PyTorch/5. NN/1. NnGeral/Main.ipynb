# 1. nn.Linear

São os pesos que nosso modelo vai ter uma matriz de tamanho (input, output) é criada, cada coluna representa um neurônio ou seja (dados de entrada, neuronios)

```python
import torch
import torch.nn as nn

x = torch.rand(2, 2)
weights = nn.Linear(2, 10)
z = weights(x)

print(f'Original Data:\n{x}\n')
print(f'Matrix of weights:\n{weights}\n')
print(f'Transformed data:\n{z}')
```
