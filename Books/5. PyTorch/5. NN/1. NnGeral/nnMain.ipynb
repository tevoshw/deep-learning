# Geral `.nn`

Quando você usa nn.Linear, você está criando um objeto. Esse objeto guarda coisas dentro dele (os pesos e o viés).
- Pense assim: É como uma gaveta. Você coloca o nn.Linear no seu modelo e ele guarda os pesos lá dentro. Por isso, quando você pede model.parameters(), o PyTorch sabe onde procurar.
- Ideal para: Quando você precisa que o PyTorch lembre de algo (pesos, bias, estatísticas de uma BatchNorm).

### (1) `nn.Linear()`


Simula os pesos de uma camada inteira. É a implementação de uma Transformação Afim. Ele encapsula a lógica de multiplicação matricial entre o vetor de entrada e uma matriz de pesos treináveis. tem o shape de (input, output), aonde o input representa o total de samples que iremos receber e o output são os números de neurônios       

```python
import torch.nn as nn
import torch

x = torch.rand(2, 2)
layer1 = nn.Linear(2, 5)
z = layer1(x)

print(f'Data input (x):\n{x}\n')
print(f'Weights of the layer: \n{layer1.state_dict()}\n')
print(f'Output x @ layer1.T: \n{z}\n')
```

### (2) `nn.Sequential()`

Aqui podemos acumular modulos em uma sequencia, como blocos, agrupa várias camadas em uma sequência fixa. Quando você passa um dado por ele, o dado entra na primeira camada, o resultado vai para a segunda, depois para a terceira, e assim por diante.

```python
modules_block1 = nn.Sequential(
    nn.Linear(1, 1), # layer 1 block 1
    nn.Linear(1, 1) # layer 1 block 1
)

modules_block1 = nn.Sequential(
    nn.Linear(1, 1), # layer 1 block 2
    nn.Linear(1, 1) # layer 2 block 2
)
```

### (3) `nn.ModuleList()`

Função utilizada para repetir um bloco, sequencia ou etc

```python
x = torch.rand(2,2)
layers = nn.ModuleList(nn.Linear(2,2) for _ in range(4))

for a, layer in enumerate(layers):
    x = layer(x)
    print(f'Layer: {a}, value:\n{x}')


state = layers.state_dict().keys()
print(state)
 
```
