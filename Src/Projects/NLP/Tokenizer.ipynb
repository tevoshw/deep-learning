# 1. Get the data

```python
from datasets import load_dataset

# Get the dataset
dataset = load_dataset("imdb")
dataset1 = dataset["train"][0]['text']

print(dataset1)
```

# 2. Tokenization Area}

```python
from transformers import AutoTokenizer

# Get the tokenizer from GPT2
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Tokenizer the IMDB TEXT
tokens = tokenizer(dataset1, return_tensors='pt')


print(tokens)
```

```python
# To have the ids, we can put
print(tokens.input_ids)
```
